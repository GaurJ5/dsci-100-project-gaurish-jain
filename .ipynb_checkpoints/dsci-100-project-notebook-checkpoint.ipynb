#### Task 3.1: Tidy the data (5 points)

```r
# Select the columns that will be needed for my e research question
march_madness_tidy <- march_madness |>
  select(seed, tourney_wins)

# Remove missing values and seed 0
march_madness_tidy <- march_madness_tidy |>
  filter(!is.na(seed), !is.na(tourney_wins), seed != 0)

# Create a categorical variable
march_madness_tidy <- march_madness_tidy |>
  mutate(seed_category = as.factor(seed))

# Display the first 6 rows
head(march_madness_tidy)
```

#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidyverse)
library(tidymodels)

# Select the variables needed for the analysis
model_data <- march_madness_tidy |>
  select(seed, tourney_wins) |>
  drop_na()

# Split the data into training and testing sets
set.seed(123)

data_split <- initial_split(model_data, prop = 0.75)

training_data <- training(data_split)
testing_data <- testing(data_split)

# Create the linear regression model
linear_spec <- linear_reg() |>
  set_engine("lm")

linear_workflow <- workflow() |>
  add_model(linear_spec) |>
  add_formula(tourney_wins ~ seed)

# Fit the model using the training data
linear_fit <- fit(linear_workflow, data = training_data)

# Make predictions using the testing data
predictions <- predict(linear_fit, new_data = testing_data) |>
  bind_cols(testing_data)

# Display the actual and predicted values
predictions

# Check model performance using RMSE
model_results <- predictions |>
  metrics(truth = tourney_wins, estimate = .pred)

# Display the model performance
model_results
```

The scatterplot shows the actual number of tournament wins and the number of wins predicted by the model. Each dot represents one team from the testing dataset. The horizontal axis shows actual wins, while the vertical axis shows predicted wins.

This graph helps us understand how well the model makes predictions. When a dot is close to the line where actual and predicted values are equal, the prediction is more accurate. When a dot is far from the line, the prediction is less accurate. This difference is called an error, or residual.

Looking at the dots helps us see whether the model's predictions are close to the actual results. It also helps us identify teams where the model made larger errors.

Overall, the scatterplot shows how well tournament seed can help predict the number of wins. We can use it together with the RMSE value to evaluate the model's accuracy.

```r
# Set the graph dimensions
options(repr.plot.width = 8, repr.plot.height = 5)

# Compare actual and predicted tournament wins
ggplot(predictions, aes(x = tourney_wins, y = .pred)) +
  geom_point(colour = "steelblue", size = 2) +
  geom_abline(intercept = 0, slope = 1, colour = "red") +
  labs(
    title = "Actual vs Predicted Tournament Wins",
    x = "Actual Tournament Wins",
    y = "Predicted Tournament Wins"
  ) +
  theme(
    plot.title = element_text(hjust = 0.5, face = "bold"),
    axis.title = element_text(face = "bold")
  )
```
